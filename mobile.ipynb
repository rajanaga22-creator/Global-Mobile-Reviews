```python
# Remove currency symbols and thousands separators, then convert to numbers
local_numeric = (
    mobile_data["price_local"]
    .str.replace(r"[^\d.,-]", "", regex=True)
    .str.replace(",", "", regex=False)
)

mobile_data["price_local_numeric"] = pd.to_numeric(local_numeric, errors="coerce")

# Convert USD price to local currency for rows where local price is missing
converted_price = mobile_data["price_usd"] * mobile_data["exchange_rate_to_usd"]
mobile_data["price_local_numeric"] = mobile_data["price_local_numeric"].fillna(converted_price)

# Fill any remaining missing values with the median for that currency
currency_medians = mobile_data.groupby("currency")["price_local_numeric"].transform("median")
mobile_data["price_local_numeric"] = mobile_data["price_local_numeric"].fillna(currency_medians)

print("Remaining nulls:", mobile_data["price_local_numeric"].isna().sum())
```

```python
mobile_data["rating_was_observed"] = mobile_data["rating"].notna()

model_medians = mobile_data.groupby("model")["rating"].transform("median")
overall_median = mobile_data["rating"].median()

mobile_data["rating"] = (
    mobile_data["rating"]
    .fillna(model_medians)
    .fillna(overall_median)
    .round()
    .astype(int)
)

print("Remaining nulls:", mobile_data["rating"].isna().sum())
```

```python
most_common_source = mobile_data["source"].mode()[0]
mobile_data["source"] = mobile_data["source"].fillna(most_common_source)

print("Remaining nulls:", mobile_data["source"].isna().sum())
```

```python
rating_to_sentiment = {
    1: "Negative",
    2: "Negative",
    3: "Neutral",
    4: "Positive",
    5: "Positive"
}

missing_sentiment = mobile_data["sentiment"].isna()
mobile_data.loc[missing_sentiment, "sentiment"] = (
    mobile_data.loc[missing_sentiment, "rating"]
    .map(rating_to_sentiment)
    .fillna("Unknown")  # used when rating is also missing or invalid
)

print("Remaining nulls:", mobile_data["sentiment"].isna().sum())
```

```python
import pandas as pd

# Convert local price text (currency symbols) to a numeric value
price_text = (
    mobile_data["price_local"]
    .str.replace(r"[^\d.,-]", "", regex=True)
    .str.replace(",", "", regex=False)
)
mobile_data["price_local_numeric"] = pd.to_numeric(price_text, errors="coerce")

# Fill missing local prices using USD conversion, then the median for that currency
converted_price = mobile_data["price_usd"] * mobile_data["exchange_rate_to_usd"]
mobile_data["price_local_numeric"] = mobile_data["price_local_numeric"].fillna(converted_price)
currency_medians = mobile_data.groupby("currency")["price_local_numeric"].transform("median")
mobile_data["price_local_numeric"] = mobile_data["price_local_numeric"].fillna(currency_medians)
print("Remaining price_local_numeric nulls:", mobile_data["price_local_numeric"].isna().sum())

# Parse review dates. The sample dates appear month-first; verify this with your data source.
mobile_data["review_date"] = pd.to_datetime(
    mobile_data["review_date"],
    format="mixed",
    dayfirst=False,
    errors="coerce"
)

# Create date-based features
mobile_data["review_year"] = mobile_data["review_date"].dt.year
mobile_data["review_month"] = mobile_data["review_date"].dt.month

# Create an average of the phone feature ratings
feature_cols = [
    "battery_life_rating",
    "camera_rating",
    "performance_rating",
    "design_rating",
    "display_rating"
]
mobile_data["average_feature_rating"] = mobile_data[feature_cols].mean(axis=1)

# Optional age groups
mobile_data["age_group"] = pd.cut(
    mobile_data["age"],
    bins=[0, 24, 34, 44, 54, float("inf")],
    labels=["18–24", "25–34", "35–44", "45–54", "55+"]
)

# Check the new columns and any dates that did not parse
print(mobile_data[[
    "price_local_numeric", "review_date", "review_year",
    "review_month", "average_feature_rating", "age_group"
]].head())

print("Unparsed dates:", mobile_data["review_date"].isna().sum())
```

```python
from sklearn.cluster import KMeans
from sklearn.impute import SimpleImputer
from sklearn.metrics import silhouette_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# --- Customer/review segmentation ---
customer_features = mobile_data[
    [
        "age",
        "rating",
        "battery_life_rating",
        "camera_rating",
        "performance_rating",
        "design_rating",
        "display_rating",
        "helpful_votes",
    ]
].copy()

# Include whether the purchase was verified as a numeric feature
customer_features["verified_purchase"] = (
    mobile_data["verified_purchase"].astype(int)
)

customer_clusterer = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    KMeans(n_clusters=4, random_state=42, n_init=50),
)

mobile_data["customer_segment"] = customer_clusterer.fit_predict(
    customer_features
)

# Score the customer/review clustering in the scaled feature space
scaled_customer_features = customer_clusterer[:-1].transform(customer_features)
customer_silhouette = silhouette_score(
    scaled_customer_features,
    mobile_data["customer_segment"],
    sample_size=5000,
    random_state=42,
)
print(f"Customer silhouette score: {customer_silhouette:.3f}")

# Aggregate reviews into customer profiles for stronger, more meaningful clustering
customer_profiles = mobile_data.groupby("customer_name", as_index=False).agg(
    average_age=("age", "mean"),
    average_rating=("rating", "mean"),
    average_helpful_votes=("helpful_votes", "mean"),
    average_feature_rating=("average_feature_rating", "mean"),
    verified_purchase_ratio=("verified_purchase", "mean"),
)
profile_features = customer_profiles[
    [
        "average_age",
        "average_rating",
        "average_helpful_votes",
        "average_feature_rating",
        "verified_purchase_ratio",
    ]
]
profile_scores = {}
for k in range(2, min(11, len(customer_profiles))):
    profile_clusterer = make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        KMeans(n_clusters=k, random_state=42, n_init=50),
    )
    profile_labels = profile_clusterer.fit_predict(profile_features)
    profile_scaled = profile_clusterer[:-1].transform(profile_features)
    profile_scores[k] = silhouette_score(profile_scaled, profile_labels)
best_profile_k = max(profile_scores, key=profile_scores.get)
print(
    f"Aggregated customer silhouette score: "
    f"{profile_scores[best_profile_k]:.3f} with k={best_profile_k}"
)

# Review/feature profiles for each segment
customer_summary = mobile_data.groupby("customer_segment")[
    ["age", "rating", "helpful_votes", "battery_life_rating",
     "camera_rating", "performance_rating", "design_rating",
     "display_rating"]
].mean().round(2)

display(customer_summary)
```

```python
# --- Product segmentation ---
# First summarize reviews into one row per phone model
products = (
    mobile_data.groupby(["brand", "model"], as_index=False)
    .agg(
        review_count=("review_id", "count"),
        avg_price_usd=("price_usd", "mean"),
        avg_rating=("rating", "mean"),
        battery_rating=("battery_life_rating", "mean"),
        camera_rating=("camera_rating", "mean"),
        performance_rating=("performance_rating", "mean"),
        design_rating=("design_rating", "mean"),
        display_rating=("display_rating", "mean"),
    )
)

product_features = products[
    [
        "review_count",
        "avg_price_usd",
        "avg_rating",
        "battery_rating",
        "camera_rating",
        "performance_rating",
        "design_rating",
        "display_rating",
    ]
]

product_clusterer = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    KMeans(n_clusters=4, random_state=42, n_init=50),
)

products["product_segment"] = product_clusterer.fit_predict(product_features)

# Score the product clustering in the scaled feature space
scaled_product_features = product_clusterer[:-1].transform(product_features)
product_silhouette = silhouette_score(
    scaled_product_features,
    products["product_segment"],
)
print(f"Product silhouette score: {product_silhouette:.3f}")

# Compare feature subsets and cluster counts to find a stronger configuration
products["average_feature_rating"] = products[
    ["battery_rating", "camera_rating", "performance_rating",
     "design_rating", "display_rating"]
].mean(axis=1)
product_feature_sets = {
    "all": ["review_count", "avg_price_usd", "avg_rating",
            "battery_rating", "camera_rating", "performance_rating",
            "design_rating", "display_rating"],
    "no_review_count": ["avg_price_usd", "avg_rating",
                        "battery_rating", "camera_rating", "performance_rating",
                        "design_rating", "display_rating"],
    "core": ["avg_price_usd", "avg_rating", "average_feature_rating"],
    "price_rating": ["avg_price_usd", "avg_rating"],
}
best_product_score = -1
best_product_config = None
for feature_name, feature_columns in product_feature_sets.items():
    tuning_features = products[feature_columns]
    for k in range(2, min(11, len(products))):
        tuning_clusterer = make_pipeline(
            SimpleImputer(strategy="median"),
            StandardScaler(),
            KMeans(n_clusters=k, random_state=42, n_init=50),
        )
        tuning_labels = tuning_clusterer.fit_predict(tuning_features)
        tuning_scaled = tuning_clusterer[:-1].transform(tuning_features)
        tuning_score = silhouette_score(tuning_scaled, tuning_labels)
        if tuning_score > best_product_score:
            best_product_score = tuning_score
            best_product_config = (feature_name, k)
print(
    f"Best product configuration: {best_product_config}, "
    f"silhouette={best_product_score:.3f}"
)

display(products.sort_values(["product_segment", "brand", "model"]))
```

## Feature Scaling & Encoding Techniques

Scaling puts numeric features on comparable scales. `StandardScaler` centers values around 0 with a standard deviation of 1; `MinMaxScaler` rescales them to 0-1. One-hot encoding converts categories into binary columns without imposing a false numeric order. These examples create separate transformed data and leave `mobile_data` unchanged. For predictive modeling, split the data first and fit transformations on the training set only.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

numeric_features = [
    "age",
    "price_usd",
    "rating",
    "battery_life_rating",
    "camera_rating",
    "performance_rating",
    "design_rating",
    "display_rating",
    "helpful_votes",
]
numeric_data = mobile_data[numeric_features]

standard_scaled = pd.DataFrame(
    StandardScaler().fit_transform(numeric_data),
    columns=numeric_features,
    index=mobile_data.index,
)
minmax_scaled = pd.DataFrame(
    MinMaxScaler().fit_transform(numeric_data),
    columns=numeric_features,
    index=mobile_data.index,
)

display(standard_scaled.head())
display(minmax_scaled.head())
```

## Similarity-Based Recommendation System

This content-based example compares phone models by price, average rating, and feature ratings. Standardization prevents price scale from dominating, and cosine similarity ranks phones with similar profiles. It recommends similar products, not necessarily products a particular person will like.

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import StandardScaler

recommendation_features = [
    "avg_price_usd",
    "avg_rating",
    "battery_rating",
    "camera_rating",
    "performance_rating",
    "design_rating",
    "display_rating",
]
products["product_label"] = products["brand"] + " " + products["model"]
scaled_recommendation_features = StandardScaler().fit_transform(
    products[recommendation_features]
)
phone_similarity = pd.DataFrame(
    cosine_similarity(scaled_recommendation_features),
    index=products["product_label"],
    columns=products["product_label"],
)

target_product = products.loc[
    products["model"].eq("iPhone 15 Pro"), "product_label"
].iloc[0]
similarity_scores = phone_similarity[target_product].drop(target_product)
top_similar_phones = similarity_scores.nlargest(5).rename("cosine_similarity")
display(
    products.set_index("product_label")
    .loc[top_similar_phones.index, ["brand", "model", "avg_price_usd", "avg_rating"]]
    .join(top_similar_phones)
)
```

## Collaborative Filtering Basics

Collaborative filtering recommends items from patterns in user-item interactions rather than product attributes. The item-item example below builds a customer-by-phone rating matrix from observed ratings, computes phone similarity from overlapping rating patterns, and ranks unseen phones for a customer with the richest history. The matrix is sparse, so these results are only a teaching example; customers with no history need a popularity or content-based fallback.

```python
from sklearn.metrics.pairwise import cosine_similarity

# Exclude ratings that were filled during preprocessing; they are not user feedback.
observed_ratings = mobile_data.loc[
    mobile_data["rating_was_observed"],
    ["customer_name", "model", "rating"],
].dropna()
user_item_ratings = observed_ratings.pivot_table(
    index="customer_name",
    columns="model",
    values="rating",
    aggfunc="mean",
)
filled_user_item = user_item_ratings.fillna(0)
item_similarity_values = cosine_similarity(filled_user_item.T)
item_similarity = pd.DataFrame(
    item_similarity_values,
    index=user_item_ratings.columns,
    columns=user_item_ratings.columns,
)

rated_item_counts = user_item_ratings.notna().sum(axis=1)
demo_user = rated_item_counts.idxmax()
known_ratings = user_item_ratings.loc[demo_user].dropna()
centered_ratings = known_ratings - known_ratings.mean()
similarities_to_rated = item_similarity.loc[:, known_ratings.index]
score_denominator = similarities_to_rated.abs().sum(axis=1).replace(0, float("nan"))
collaborative_scores = (
    similarities_to_rated.mul(centered_ratings, axis="columns").sum(axis=1)
    / score_denominator
)
collaborative_scores.loc[known_ratings.index] = float("nan")

print("Customers with ratings for multiple phones:", int(rated_item_counts.gt(1).sum()))
print(
    "User-item matrix sparsity:",
    f"{1 - user_item_ratings.notna().to_numpy().mean():.1%}",
)
display(
    collaborative_scores.dropna()
    .nlargest(5)
    .rename("collaborative_score")
    .to_frame()
)
```
